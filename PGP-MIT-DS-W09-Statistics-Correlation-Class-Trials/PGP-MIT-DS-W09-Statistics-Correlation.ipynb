```python
###🧪 Python Code (Functions + Manual Calculation)
import numpy as np

# -----------------------------
# Sample Data
# -----------------------------
X = np.array([1, 2, 3, 4, 5])
Y = np.array([2, 4, 6, 8, 10])

n = len(X)

# -----------------------------
# 1. Using Built-in Functions
# -----------------------------
cov_matrix = np.cov(X, Y)
cov_function = cov_matrix[0, 1]

corr_matrix = np.corrcoef(X, Y)
corr_function = corr_matrix[0, 1]

# -----------------------------
# 2. Manual Calculation
# -----------------------------

# Means
mean_x = np.mean(X)
mean_y = np.mean(Y)

# Deviations
dev_x = X - mean_x
dev_y = Y - mean_y

# ---- Covariance (manual) ----
cov_manual = np.sum(dev_x * dev_y) / (n - 1)

# ---- Standard Deviations (manual) ----
std_x = np.sqrt(np.sum(dev_x**2) / (n - 1))
std_y = np.sqrt(np.sum(dev_y**2) / (n - 1))

# ---- Correlation (manual) ----
corr_manual = cov_manual / (std_x * std_y)

# -----------------------------
# 3. Output Results
# -----------------------------
print("=== USING NUMPY FUNCTIONS ===")
print("Covariance:", cov_function)
print("Correlation:", corr_function)

print("\n=== MANUAL CALCULATION ===")
print("Mean X:", mean_x)
print("Mean Y:", mean_y)
print("Std Dev X:", std_x)
print("Std Dev Y:", std_y)
print("Covariance (manual):", cov_manual)
```
